## 1. Load and Inspect the Dataset

```python
import pandas as pd

# Load the raw student performance dataset
file_path = "../data/student-mat.csv"
df = pd.read_csv(file_path, sep=";")

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
# Display the first five rows of the dataset
df.head()
```

```python
# Display dataset information
df.info()
```

```python
# Generate summary statistics for numerical columns
df.describe()
```

```python
# Summary statistics for categorical columns
df.describe(include="object")
```

## 2. Data Quality Analysis

```python
# Check the number of missing values in each column
missing_values = df.isnull().sum()

print("Missing values by column:")
print(missing_values)
```

```python
# Calculate the total number of missing values
total_missing = df.isnull().sum().sum()

print("Total missing values:", total_missing)
```

```python
# Check for duplicate records
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Display the data type of each column
df.dtypes
```

```python
# Count columns by data type
df.dtypes.value_counts()
```

```python
# Inspect unique values in selected categorical columns
categorical_columns = [
    "school",
    "sex",
    "address",
    "famsize",
    "Pstatus",
    "Mjob",
    "Fjob"
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
# Check the number of unique values in every column
unique_counts = df.nunique().sort_values()

print(unique_counts)
```

```python
# Select numerical columns
numeric_columns = df.select_dtypes(include="number").columns

print("Numerical columns:")
print(list(numeric_columns))
```

```python
# Check minimum and maximum values for numerical columns
numeric_range = df[numeric_columns].agg(["min", "max"]).T

print(numeric_range)
```

```python
# Check for negative values in numerical columns
negative_counts = (df[numeric_columns] < 0).sum()

print("Negative values by numerical column:")
print(negative_counts)
```

## 3. Data Cleaning

```python
# Create a separate copy for data cleaning
cleaned_df = df.copy()

print("Cleaned dataset copy created.")
print("Original shape:", df.shape)
print("Cleaning dataset shape:", cleaned_df.shape)
```

```python
# Handle missing values
missing_before = cleaned_df.isnull().sum().sum()

if missing_before > 0:
    cleaned_df = cleaned_df.dropna()
    print(f"Missing values found and removed: {missing_before}")
else:
    print("No missing values found. Dataset does not require missing-value removal.")

missing_after = cleaned_df.isnull().sum().sum()

print("Missing values after cleaning:", missing_after)
```

```python
# Handle duplicate records
duplicates_before = cleaned_df.duplicated().sum()

if duplicates_before > 0:
    cleaned_df = cleaned_df.drop_duplicates()
    print(f"Duplicate rows removed: {duplicates_before}")
else:
    print("No duplicate records found. No rows were removed.")

duplicates_after = cleaned_df.duplicated().sum()

print("Duplicate rows after cleaning:", duplicates_after)
```

```python
# Final data quality validation

print("FINAL DATA QUALITY CHECK")
print("-" * 30)

print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_df.duplicated().sum())
```

```python
# Verify data types after cleaning

print("Data types after cleaning:")
print(cleaned_df.dtypes)
```

```python
# Export the cleaned dataset
output_path = "../data/student-mat-cleaned.csv"

cleaned_df.to_csv(output_path, index=False)

print("Cleaned dataset exported successfully.")
print("Saved to:", output_path)
```

```python
# Compare the original and cleaned datasets

print("DATASET COMPARISON")
print("-" * 35)

print("Original dataset:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nCleaned dataset:")
print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])

print("\nChanges:")
print("Rows removed:", df.shape[0] - cleaned_df.shape[0])
print("Columns removed:", df.shape[1] - cleaned_df.shape[1])
```

```python
# Verify the exported cleaned dataset

verified_df = pd.read_csv("../data/student-mat-cleaned.csv")

print("Exported file verified successfully.")
print("Shape:", verified_df.shape)
print("Missing values:", verified_df.isnull().sum().sum())
print("Duplicate rows:", verified_df.duplicated().sum())
```

## 4. Basic Data Analysis

```python
# Calculate average grades

grade_columns = ["G1", "G2", "G3"]

average_grades = cleaned_df[grade_columns].mean()

print("Average Student Grades")
print("-" * 25)

for grade, average in average_grades.items():
    print(f"{grade}: {average:.2f}")
```

```python
# Analyze the final grade (G3)

print("Final Grade Analysis")
print("-" * 25)
print("Average:", round(cleaned_df["G3"].mean(), 2))
print("Minimum:", cleaned_df["G3"].min())
print("Maximum:", cleaned_df["G3"].max())
print("Median:", cleaned_df["G3"].median())
```

## 5. Data Visualization

```python
import matplotlib.pyplot as plt

# Visualize the distribution of final grades
plt.figure(figsize=(8, 5))

plt.hist(cleaned_df["G3"], bins=11, edgecolor="black")

plt.title("Distribution of Final Mathematics Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
# Compare average final grades by gender

gender_analysis = (
    cleaned_df.groupby("sex")["G3"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

print("Average Final Grade by Gender")
print("-" * 30)
print(gender_analysis)
```
