```python
# DevSphere AI & ML Internship - Week 2
# Project: Student Performance Data Visualization

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

# Set a clean visualization style
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

```python
# Load the student performance dataset

file_path = "../data/student-mat.csv"

df = pd.read_csv(file_path, sep=";")

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)

df.head()
```

# Student Performance Data Visualization

This notebook explores the Student Performance dataset through clear and meaningful visualizations. Matplotlib and Seaborn are used to create bar, line, and pie charts that help us understand student performance and identify useful patterns in the data.

```python
# Bar Chart: Average Final Grade by Gender

average_grade_gender = df.groupby("sex")["G3"].mean()

plt.figure(figsize=(8, 5))
sns.barplot(
    x=average_grade_gender.index,
    y=average_grade_gender.values
)

plt.title("Average Final Grade (G3) by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")
plt.tight_layout()

plt.show()
```

```python
# Line Chart: Average Grades Across G1, G2, and G3

average_grades = df[["G1", "G2", "G3"]].mean()

plt.figure(figsize=(8, 5))
plt.plot(
    average_grades.index,
    average_grades.values,
    marker="o",
    linewidth=2
)

plt.title("Average Student Grades Across G1, G2, and G3")
plt.xlabel("Grade Period")
plt.ylabel("Average Grade")
plt.ylim(0, 20)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
# Pie Chart: Student Distribution by Final Grade Performance

def performance_group(grade):
    if grade >= 15:
        return "High (15-20)"
    elif grade >= 10:
        return "Medium (10-14)"
    else:
        return "Low (0-9)"

performance_groups = df["G3"].apply(performance_group).value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    performance_groups.values,
    labels=performance_groups.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Distribution by Final Grade Performance")
plt.tight_layout()

plt.show()
```

## Key Insights

### 1. Average Final Grade by Gender
The bar chart shows a small difference in the average final grades between female and male students. This suggests that gender alone does not create a major difference in overall academic performance in this dataset.

### 2. Average Grades Across G1, G2, and G3
The line chart shows how average grades change from the first period (G1) to the second period (G2) and finally to the final grade (G3). The values help us understand the overall academic performance trend of the students.

### 3. Student Distribution by Final Grade Performance
The pie chart shows the proportion of students in high, medium, and low final-grade groups. The largest group represents the most common performance level among the students, giving a quick overview of the overall distribution.

### Overall Observation
The visualizations make the student performance data easier to understand than raw tables alone. They highlight differences, trends, and performance distributions in a simple and clear way.
